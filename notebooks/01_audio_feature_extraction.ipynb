## EXTRACCION DE DATOS
Para el desarrollo de este proyecto se recurrió a dos tipos de fuentes. Por un lado, las pistas de audio utilizadas para construir el dataset fueron obtenidas de Free Music Archive (https://freemusicarchive.org), plataforma de distribución musical que ofrece canciones bajo licencias Creative Commons, permitiendo su uso libre en contextos académicos e investigativos sin restricciones de derechos de autor. Las canciones fueron seleccionadas y organizadas manualmente por género musical, garantizando un mínimo de 50 pistas por categoría.

## PROCESAMIENTO Y MODELAMIENTO DE LOS DATOS

### Librerias
Librosa: es una librería especializada en análisis de audio y música que permite cargar archivos de sonido y extraer características acústicas. Es el núcleo del proceso de extracción de features en este proyecto.

NumPy: es la librería fundamental para computación numérica en Python. Proporciona estructuras de datos tipo array de alto rendimiento y operaciones matemáticas vectorizadas, siendo la base sobre la que trabajan la mayoría de librerías científicas incluyendo Librosa.

Pandas es la librería estándar para manipulación y análisis de datos tabulares en Python. Permite construir, limpiar y transformar dataframes, que es la estructura utilizada para almacenar y exportar el dataset final en formato CSV.

Os es un módulo nativo de Python que permite interactuar con el sistema operativo, específicamente con el sistema de archivos. En este proyecto se utiliza para recorrer las carpetas de géneros, listar los archivos de audio disponibles y construir las rutas de acceso a cada canción.

### Extracción de Features
El código implementa un pipeline de extracción automática de características acústicas a partir de archivos de audio organizados por género musical. Para cada canción se carga la pista completa y se extraen segmentos de audio. De cada segmento se calculan 47 features acústicas:

**11 features globales:** tempo, spectral_centroid, spectral_bandwidth, rolloff, zero_crossing_rate, rms, chroma_stft, onset_strength_mean, onset_strength_std, pitch_variance, spectral_flux_std

**26 coeficientes MFCC:** mfcc1 a mfcc13, cada uno con su media y desviación estándar (mean + std)

**10 features avanzadas:** harm_perc_ratio (energía armónica/percusiva), spectral_flatness_mean, tonnetz_0_std a tonnetz_5_std (6 coeficientes tonales), tempo_cv (variabilidad del tempo), bass_ratio (ratio graves/agudos)

Cada segmento procesado queda registrado como una fila independiente en el dataset, junto con su etiqueta de género y el nombre del archivo de origen. El resultado final es un archivo CSV estructurado con 50 columnas (47 features + 3 metadatos: label, song_id, segment_type) y listo para la etapa de modelado.

## Librerías y función de features compartida

> Nota: Este notebook requiere FFmpeg para decodificar archivos MP3. Si no lo tienes instalado, ejecuta en PowerShell: `winget install ffmpeg`. Una vez instalado, reinicia VSCode completamente para que el PATH se actualice y los workers de joblib puedan encontrarlo.

```python
import librosa
import numpy as np
import pandas as pd
import os
import random
import warnings
from joblib import Parallel, delayed
from tqdm import tqdm

warnings.filterwarnings('ignore')

random.seed(42)


def extraerfeatures(y, sr):
    mfccs     = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
    onset_env = librosa.onset.onset_strength(y=y, sr=sr)
    chroma    = librosa.feature.chroma_stft(y=y, sr=sr)
    y_harm, y_perc = librosa.effects.hpss(y)

    features = {
        'tempo':               float(librosa.beat.tempo(y=y, sr=sr)[0]),
        'spectral_centroid':   float(librosa.feature.spectral_centroid(y=y, sr=sr).mean()),
        'spectral_bandwidth':  float(librosa.feature.spectral_bandwidth(y=y, sr=sr).mean()),
        'rolloff':             float(librosa.feature.spectral_rolloff(y=y, sr=sr).mean()),
        'zero_crossing_rate':  float(librosa.feature.zero_crossing_rate(y).mean()),
        'rms':                 float(librosa.feature.rms(y=y).mean()),
        'chroma_stft':         float(chroma.mean()),
        'onset_strength_mean': float(onset_env.mean()),
        'onset_strength_std':  float(onset_env.std()),
        'pitch_variance':      float(chroma.var()),
        'spectral_flux_std':   float(np.std(np.sum(np.diff(np.abs(librosa.stft(y)), axis=1)**2, axis=0))),
    }

    for i, coef in enumerate(mfccs):
        features[f'mfcc{i + 1}_mean'] = float(coef.mean())
        features[f'mfcc{i + 1}_std']  = float(coef.std())

    harm_energy = float(np.mean(np.abs(y_harm)))
    perc_energy = float(np.mean(np.abs(y_perc)))
    features['harm_perc_ratio'] = harm_energy / (perc_energy + 1e-6)

    features['spectral_flatness_mean'] = float(
        librosa.feature.spectral_flatness(y=y).mean()
    )

    tonnetz = librosa.feature.tonnetz(y=y_harm, sr=sr)
    for i in range(tonnetz.shape[0]):
        features[f'tonnetz_{i}_std'] = float(tonnetz[i].std())

    _, beats = librosa.beat.beat_track(y=y, sr=sr)
    if len(beats) > 1:
        intervals = np.diff(librosa.frames_to_time(beats, sr=sr))
        features['tempo_cv'] = float(
            np.std(intervals) / (np.mean(intervals) + 1e-6)
        )
    else:
        features['tempo_cv'] = 0.0

    stft  = np.abs(librosa.stft(y))
    freqs = librosa.fft_frequencies(sr=sr)
    bass = stft[freqs < 250]
    treble = stft[freqs > 2000]
    if len(treble) == 0:
        features['bass_ratio'] = float(np.mean(bass)) if len(bass) > 0 else 0.0
    else:
        features['bass_ratio'] = float(
            np.mean(bass) / (np.mean(treble) + 1e-6)
        )

    return features


print('Librerías cargadas OK')
```

## CODIGO PARA SEGMENTOS DE 30s

```python
def procesar_cancion_30s(ruta, genero, archivo):
    """Procesa una canción y devuelve sus segmentos de 30s."""
    filas = []

    try:
        y, sr = librosa.load(ruta, sr=None)
        duracion = librosa.get_duration(y=y, sr=sr)

        if duracion >= 85:
            puntos = [
                (0, 'inicio'),
                (int(duracion / 2) - 15, 'mitad'),
                (int(duracion) - 30, 'final'),
            ]
        elif duracion >= 60:
            puntos = [(0, 'inicio'), (int(duracion) - 30, 'final')]
        elif duracion >= 30:
            puntos = [(0, 'unico')]
        else:
            return [], 'corto', None, None, None

        for inicio_seg, nombre_seg in puntos:
            inicio_p = int(inicio_seg * sr)
            fin_p    = int((inicio_seg + 30) * sr)
            segmento = y[inicio_p:fin_p]

            if len(segmento) >= int(29 * sr):
                f = extraerfeatures(segmento, sr)
                f['label']        = genero
                f['song_id']      = archivo
                f['segment_type'] = nombre_seg
                filas.append(f)

    except Exception as e:
        return [], 'error', archivo, type(e).__name__, str(e)

    return filas, 'ok', None, None, None
```

```python
import time

CARPETA_RAIZ_30S = '../Canciones/'
N_JOBS = -1

tareas_30s = []
generos_30s = [
    d for d in os.listdir(CARPETA_RAIZ_30S)
    if os.path.isdir(os.path.join(CARPETA_RAIZ_30S, d))
]

for genero in generos_30s:
    carpeta = os.path.join(CARPETA_RAIZ_30S, genero)
    for archivo in os.listdir(carpeta):
        if archivo.endswith(('.mp3', '.wav')):
            tareas_30s.append((os.path.join(carpeta, archivo), genero, archivo))

print(f'Total canciones a procesar: {len(tareas_30s)}')
print(f'Géneros: {generos_30s}')
print(f'Iniciando con n_jobs={N_JOBS}...\n')

t0 = time.time()

resultados_30s = Parallel(n_jobs=N_JOBS, backend='loky', verbose=5)(
    delayed(procesar_cancion_30s)(ruta, genero, archivo)
    for ruta, genero, archivo in tareas_30s
)

t1 = time.time()

filas_30s = []
stats_30s = {'procesados': 0, 'error': 0, 'corto': 0}
errores_30s = []  # (archivo, tipo_error, mensaje)

for filas, estado, archivo_err, tipo_err, msg_err in resultados_30s:
    if estado == 'ok':
        stats_30s['procesados'] += 1
        filas_30s.extend(filas)
    elif estado == 'error':
        stats_30s['error'] += 1
        errores_30s.append((archivo_err, tipo_err, msg_err))
    elif estado == 'corto':
        stats_30s['corto'] += 1

df_30s = pd.DataFrame(filas_30s)
df_30s.to_csv('../data/songs/audio_features_v1_30s.csv', index=False)

print(f'\n{"=" * 50}')
print('  REPORTE — ETL 30s PARALELIZADO')
print(f'{"=" * 50}')
print(f'  Tiempo total:              {t1 - t0:.1f}s ({(t1-t0)/60:.1f} min)')
print(f'  Canciones procesadas:      {stats_30s["procesados"]}')
print(f'  Canciones con error:       {stats_30s["error"]}')
print(f'  Canciones muy cortas:      {stats_30s["corto"]}')
print(f'  Total segmentos:           {df_30s.shape[0]}')
print(f'  Características:           {df_30s.shape[1] - 3}')
print(f'\n  Distribución por género:')
for genero, count in df_30s['label'].value_counts().items():
    print(f'    {genero}: {count} segmentos')

if errores_30s:
    from collections import Counter
    conteo = Counter(tipo for _, tipo, _ in errores_30s)
    print(f'\n  Errores por tipo:')
    for tipo, n in conteo.most_common():
        print(f'    {tipo}: {n} canciones')
    print(f'\n  Detalle de errores:')
    for archivo, tipo, msg in errores_30s:
        print(f'    [{tipo}] {archivo}: {msg[:80]}')

print(f'{"=" * 50}')
print('Dataset guardado en: ../data/songs/audio_features_v1_30s.csv')
```

## ETL CON VENTANAS DESLIZANTES (15s, solapamiento 1s)

Esta variación del ETL original procesa cada canción usando ventanas deslizantes
de 15 segundos con 1 segundo de solapamiento entre ventanas consecutivas
(paso de 14 segundos). A diferencia del ETL original que extraía 1 a 3
segmentos fijos por canción, este enfoque genera una secuencia densa de
segmentos que captura la evolución temporal de las características acústicas
a lo largo de toda la pista.

**Fuente de audio:** `Canciones/` (7 géneros, ~50 canciones cada uno)

**Cambios respecto al ETL original:**
- Segmentos de 15s en lugar de 30s
- Solapamiento de 1s entre ventanas consecutivas
- Todas las canciones de ≥15s se procesan (no solo las de ≥30s)
- Los segmentos se nombran como `seg_XXXX` en orden secuencial
- El archivo de salida es `../data/songs/audio_features_15s.csv`

```python
DURACION_VENTANA      = 15
SOLAPAMIENTO          = 1
PASO                  = DURACION_VENTANA - SOLAPAMIENTO  # 14s
MAX_SEGS_POR_CANCION  = 15


def procesar_cancion_15s(ruta, genero, archivo):
    """Procesa una canción con ventanas deslizantes de 15s."""
    filas = []

    try:
        y, sr = librosa.load(ruta, sr=None)
        duracion = librosa.get_duration(y=y, sr=sr)

        if duracion < DURACION_VENTANA:
            return [], 'corto', None, None, None

        num_segmento = 0
        inicio = 0.0
        segmentos_temp = []

        while inicio + DURACION_VENTANA <= duracion:
            inicio_p = int(inicio * sr)
            fin_p    = int((inicio + DURACION_VENTANA) * sr)
            segmento = y[inicio_p:fin_p]

            if len(segmento) >= int(14 * sr):
                f = extraerfeatures(segmento, sr)
                f['label']        = genero
                f['song_id']      = archivo
                f['segment_type'] = f'seg_{num_segmento:04d}'
                segmentos_temp.append(f)

            inicio += PASO
            num_segmento += 1

        if len(segmentos_temp) > MAX_SEGS_POR_CANCION:
            segmentos_temp = random.sample(segmentos_temp, MAX_SEGS_POR_CANCION)

        filas.extend(segmentos_temp)

    except Exception as e:
        return [], 'error', archivo, type(e).__name__, str(e)

    return filas, 'ok', None, None, None
```

```python
CARPETA_RAIZ_15S = '../Canciones/'

tareas_15s = []
generos_15s = [
    d for d in os.listdir(CARPETA_RAIZ_15S)
    if os.path.isdir(os.path.join(CARPETA_RAIZ_15S, d))
]

for genero in generos_15s:
    carpeta = os.path.join(CARPETA_RAIZ_15S, genero)
    for archivo in os.listdir(carpeta):
        if archivo.endswith(('.mp3', '.wav')):
            tareas_15s.append((os.path.join(carpeta, archivo), genero, archivo))

print(f'Total canciones a procesar: {len(tareas_15s)}')
print(f'Géneros: {generos_15s}')
print(f'Iniciando con n_jobs={N_JOBS}...\n')

t0 = time.time()

resultados_15s = Parallel(n_jobs=N_JOBS, backend='loky', verbose=5)(
    delayed(procesar_cancion_15s)(ruta, genero, archivo)
    for ruta, genero, archivo in tareas_15s
)

t1 = time.time()

filas_15s = []
stats_15s = {'procesados': 0, 'error': 0, 'corto': 0}
errores_15s = []  # (archivo, tipo_error, mensaje)

for filas, estado, archivo_err, tipo_err, msg_err in resultados_15s:
    if estado == 'ok':
        stats_15s['procesados'] += 1
        filas_15s.extend(filas)
    elif estado == 'error':
        stats_15s['error'] += 1
        errores_15s.append((archivo_err, tipo_err, msg_err))
    elif estado == 'corto':
        stats_15s['corto'] += 1

df_15s = pd.DataFrame(filas_15s)
df_15s.to_csv('../data/songs/audio_features_15s.csv', index=False)

print(f'\n{"=" * 55}')
print('  REPORTE — ETL 15s VENTANAS DESLIZANTES PARALELIZADO')
print(f'{"=" * 55}')
print(f'  Tiempo total:              {t1 - t0:.1f}s ({(t1-t0)/60:.1f} min)')
print(f'  Ventana: {DURACION_VENTANA}s | Solapamiento: {SOLAPAMIENTO}s | Tope: {MAX_SEGS_POR_CANCION}')
print(f'  Canciones procesadas:      {stats_15s["procesados"]}')
print(f'  Canciones con error:       {stats_15s["error"]}')
print(f'  Canciones muy cortas:      {stats_15s["corto"]}')
print(f'  Total segmentos:           {df_15s.shape[0]}')
print(f'  Características:           {df_15s.shape[1] - 3}')
print(f'\n  Distribución por género:')
for genero, count in df_15s['label'].value_counts().items():
    print(f'    {genero}: {count} segmentos')

if errores_15s:
    from collections import Counter
    conteo = Counter(tipo for _, tipo, _ in errores_15s)
    print(f'\n  Errores por tipo:')
    for tipo, n in conteo.most_common():
        print(f'    {tipo}: {n} canciones')
    print(f'\n  Detalle de errores:')
    for archivo, tipo, msg in errores_15s:
        print(f'    [{tipo}] {archivo}: {msg[:80]}')

print(f'{"=" * 55}')
print('Dataset guardado en: ../data/songs/audio_features_15s.csv')
```

```python
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set(style='whitegrid')

# Preparar los datos: excluir columnas no numéricas y NaNs
X = df_15s.drop(['label', 'song_id', 'segment_type'], axis=1)
X = X.select_dtypes(include=[np.number]).fillna(0)
labels = df_15s['label'].astype(str).values

# Escalado
scaler = StandardScaler()
Xs = scaler.fit_transform(X)

# PCA previo (reduce ruido y acelera t-SNE)
n_pca = min(50, Xs.shape[1])
pca = PCA(n_components=n_pca, random_state=42)
X_pca = pca.fit_transform(Xs)

# t-SNE para mejor separación visual (inicializado con PCA)
# Evitar argumentos incompatibles entre versiones de scikit-learn
tsne = TSNE(n_components=2, perplexity=30, init='pca', random_state=42)
X_emb = tsne.fit_transform(X_pca)

plot_df = pd.DataFrame(X_emb, columns=['Dim1', 'Dim2'])
plot_df['label'] = labels

# KMeans para mostrar clusters (cantidad = número de géneros)
n_clusters = df_15s['label'].nunique()
kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10).fit(X_emb)
plot_df['cluster'] = kmeans.labels_
centers = kmeans.cluster_centers_

plt.figure(figsize=(12, 8))
palette = sns.color_palette('tab10', n_clusters)
ax = sns.scatterplot(data=plot_df, x='Dim1', y='Dim2', hue='label', palette='tab10', alpha=0.75, s=50, linewidth=0)

# Dibujar centros de cluster
for i, (cx, cy) in enumerate(centers):
    plt.scatter(cx, cy, marker='X', s=200, c='k', edgecolor='white')
    plt.text(cx, cy, str(i), color='white', ha='center', va='center', weight='bold')

plt.title('t-SNE (después de PCA) — separación y clusters (KMeans) (15s)')
plt.xlabel('Dim1')
plt.ylabel('Dim2')
plt.legend(title='Género', bbox_to_anchor=(1.05, 1), loc=2)
plt.grid(True, alpha=0.25)
plt.tight_layout()
plt.show()

print(f'PCA explicado (primeros 5 componentes): {pca.explained_variance_ratio_[:5]}')
print(f'Número de clusters (KMeans): {n_clusters}')
```
